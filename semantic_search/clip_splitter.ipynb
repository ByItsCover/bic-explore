```python
import os
import onnx
import onnxruntime as ort
from onnxruntime import InferenceSession
from transformers import CLIPImageProcessor, CLIPTokenizer
```

```python
destination = "clip_models"
save_dir = os.path.join(destination, "onnx")
text_file_name = "clip_text.onnx"

clip_text_onnx_model = onnx.load(os.path.join(save_dir, text_file_name))
clip_text_onnx_model
```

```python
from onnx.external_data_helper import convert_model_to_external_data

convert_model_to_external_data(clip_text_onnx_model, all_tensors_to_one_file=False, location="extra_stuff", size_threshold=0, convert_attribute=False)
```

```python
onnx.save(clip_text_onnx_model, os.path.join(save_dir, "clip_text_split.onnx"))
```
